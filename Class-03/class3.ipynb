```python
def calculate_area(width,height=10):
    """
    calculate the area of a rectangle &
    return the area also if height is not given then it takes default value of height as 10.

    """
    area = width * height
    return area
#using positional arguments
print("Area of rectangle : ",calculate_area(6,9))
```

```python
#call on default value of height
print("Area of rectangle : ",calculate_area(6))
```

```python
#call using keyword arguments
print("Area of rectangle : ",calculate_area(height=5,width=4))
```

```python
# Accessing function docstring
print("Docstring:", calculate_area.__doc__)
```

```python
def calculate_area(radius=10):
    """
    calculate the area of a circle &
    return the area also if radius is not given then take default value of radius as 10.

    """
    area = 3.1416* radius * radius
    return area
print("Area of circle : ",calculate_area(5))
```

```python
#using default value of radius
print("Area of circle : ",calculate_area())
```

```python
#set(unordered, unindexed, mutable, no duplicate elements)
set_a = {1,2,3,4,5,5,5,5}   #note that duplicate element 5 will be stored only once in the set
set_b = {4,5,6,7,8}
print("Set A:", set_a)
print("Set B:", set_b)
```

```python
#mathmatical operations on sets
union_set = set_a | set_b  # or set_a.union(set_b)
print("Union (A | B):", union_set) #all unique elements from both sets
```

```python
intersection_set = set_a & set_b  # or set_a.intersection(set_b)
print("Intersection (A & B):", intersection_set) #common elements in both sets
```

```python
difference_set = set_a - set_b  # or set_a.difference(set_b)
print("Difference (A - B):", difference_set) #elements in A but not in
```

```python
symmetric_difference_set = set_a ^ set_b  # or set_a.symmetric_difference(set_b)
print("Symmetric Difference (A ^ B):", symmetric_difference_set) #elements in A
```

```python
#Dictionaries (unordered, no duplicate keys)
#syntax: dictionary_name = {key1:value1, key2:value2, key3:value3}
mentor_profile = {
    "name": "Fahd",
    "role": "Mentor",
    "age": 25,
    "gender": "Male",
    "Topics": ["Python", "Machine Learning", "Data Science"]}
print("Mentor Profile:", mentor_profile)
```

```python
#List comprehension offer a shorter syntax to create a new list from an existing sequence or iterable.
#syntax: new_list = [expression for item in iterable if condition == True]


#task : Generate a list containing squares numbers from 1 to 5
#classic way using for loop
squares_loop = []
for x in range(1,6):
    squares_loop.append(x**2)
print("Squares using for loop:", squares_loop)
```

```python
#list comprehension way
squares_comp = [x**2 for x in range(1,6)]
print("Squares using list comprehension:", squares_comp)
```

```python
# List comprehension with conditional filter (only squares of odd numbers)
odd_squares_comp = [x ** 2 for x in range(1, 6) if x % 2 == 0]
print("Odd Squares (Comprehension):", odd_squares_comp)
```

1.Create a dictionary named inventory representing store inventory with items and prices: {"laptop": 1000, "mouse": 20, "keyboard": 50, "monitor": 250}.
2.Prompt the user for an item name. Check if it exists in the dictionary:
->If yes, display its price.
->If no, display a message "Item not found in inventory."

```python
#Exercise1:
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(is_prime(7))
print(is_prime(10))
print(is_prime(1))
print(is_prime(13))
```
