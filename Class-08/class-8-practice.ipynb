```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
titanic = pd.read_csv('/kaggle/input/datasets/yasserh/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
employee = pd.read_csv('/kaggle/input/datasets/tawfikelmetwally/employee-dataset/Employee.csv')
```

```python
stock_price_bit = np.array([8,10,15,14,20,12,25,28,24.30])
stock_price_eth = np.array([7,11,10,15,17,14,20,25,32])
year = np.array([2018,2019,2020,2021,2022,2023,2024,2025,2026])
```

```python
print(len(year),len(stock_price_bit),len(stock_price_eth))
```

```python
plt.plot(year,stock_price_bit, color = 'red', 
         linestyle = '-.', marker ='o',ms = 8, 
         mec = 'r',mfc ='b',alpha = 0.75, label = 'Bitcoin')
plt.plot(year, stock_price_eth, color ='b',
         ls ='--', marker = '*', ms = 7, 
         mec = 'g',alpha = 0.50, label = 'Etherum')
plt.legend()
plt.xlabel('year')
plt.ylabel('Stock Price')
plt.title('Year Wise Stock Price Graph')
plt.grid(color = 'red', ls = '--',alpha = 0.25,
         linewidth = 0.5,axis ='x')
plt.show()
```

```python
#piechart
class_count = titanic['Pclass'].value_counts().sort_index()
class_count
class_label = ['1st class','2nd class','3rd clss']

plt.pie(class_count, labels = class_label, 
        colors = ['r','g','b'],autopct = '%0.2f%%')
plt.title('Pie chart for Pclass')
plt.show()
```

```python
#Donutchart
class_count = titanic['Pclass'].value_counts().sort_index()
class_count
class_label = ['1st class','2nd class','3rd clss']

plt.pie(class_count, labels = class_label, 
        colors = ['r','g','b'],autopct = '%0.2f%%',
       wedgeprops = {'width' :0.6})
plt.title('Donut Chart for Pclass')
plt.show()
```

```python
#histogram
data = titanic['Age'].dropna()
plt.figure(figsize = (8,5))
plt.hist(data, bins = 10, alpha = 0.75, 
         edgecolor = 'blue')
plt.title('Age Histplot')
plt.xlabel('Age')
plt.ylabel('Number of passenger')
plt.show()
```

```python
#histogram--->cumulative
data = titanic['Age'].dropna()
plt.figure(figsize = (8,5))
plt.hist(data, color = 'green', bins = 15, alpha = 0.75, 
         edgecolor = 'blue',
        cumulative = True)
plt.title('Age Histplot--->Cumulative')
plt.xlabel('Age')
plt.ylabel('Number of passenger')
plt.show()
```
