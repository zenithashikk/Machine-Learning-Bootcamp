### Exercise 2: String Slicing and Statistics
Given the variable `course_info = "Machine Learning BootCamp 2026 - Instructor: Fahd"
1. Extract and print the substring `"BootCamp"` using slicing.
2. Extract and print the substring `"Fahd"` using slicing.
3. Reverse the entire `course_info` string and print the result.
4. Find the index where the substring `"Instructor"` begins.

```python
course_info = "Machine Learning BootCamp 2026 - Instructor: Fahd"
print(course_info[17:25])  # Extract "BootCamp"
print(course_info[0:16])  # Extract "Machine Learning"
print(course_info[45:50])  # Extract "Fahd"
print(course_info[::-1])  # Reverse the entire string
print(course_info.find("Instructor"))  # Find the index of "Instructor"
```
